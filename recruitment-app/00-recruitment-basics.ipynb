# The recruiting package, one small step at a time

This notebook is gentle. There is no agent here, no scoring, nothing clever.
Each cell does **one** small thing: connect to Drive, list what is in a
folder, read one file as text.

If you have never used a notebook: a cell is a grey box of code. Click it,
press **Shift + Enter**, and it runs. That is the whole skill.

## Step 1 - connect to Drive

This needs an OAuth `credentials.json` next to this notebook (see the
README for the exact Google Cloud Console steps) and
`GOOGLE_DRIVE_JD_FOLDER_ID` / `GOOGLE_DRIVE_RESUMES_FOLDER_ID` in your
`.env`. The first run opens a browser to sign in; after that, a cached
`token.json` remembers you.

```python
import os
from dotenv import load_dotenv

load_dotenv()  # reads the GOOGLE_DRIVE_*_FOLDER_ID and LITELLM_API_KEY from .env

from recruiting import DriveClient

drive = DriveClient()  # opens a browser the first time, to log in to Google
print("Connected to Drive.")
```

## Step 2 - list the JD folder

`list_files(folder_id)` gives back every file directly inside a folder,
newest-modified first.

```python
jd_folder_id = os.environ["GOOGLE_DRIVE_JD_FOLDER_ID"]
jd_files = drive.list_files(jd_folder_id)
print(len(jd_files), "file(s) in the JD folder:\n")
for f in jd_files:
    print(" -", f.name, "|", f.modified_time)
```

## Step 3 - list the resumes folder

Same call, a different folder.

```python
resumes_folder_id = os.environ["GOOGLE_DRIVE_RESUMES_FOLDER_ID"]
resume_files = drive.list_files(resumes_folder_id)
print(len(resume_files), "resume(s):\n")
for f in resume_files:
    print(" -", f.name, "|", f.modified_time)
```

## Step 4 - read one file as text

`latest(folder_id)` is the most recently modified file - the current JD,
even if someone just edited it. `fetch_text(file)` turns it into plain text,
whatever format it started in (pdf, docx, a Google Doc, ...).

```python
jd_file = drive.latest(jd_folder_id)
jd_text = drive.fetch_text(jd_file)
print("Reading:", jd_file.name, "\n")
print(jd_text[:500])
```

## That is all

You just used the whole `DriveClient`:

- `list_files(folder_id)` - what is in a folder
- `latest(folder_id)` - the most recently modified file
- `fetch_text(file)` - its content as plain text
